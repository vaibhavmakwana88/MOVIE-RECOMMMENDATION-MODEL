### 📝 Creating the Tags Column

To prepare text data for NLP tasks, we combine multiple descriptive fields into a single `tags` column:

- **Overview** – movie plot summary  
- **Genres** – cleaned genre names  
- **Tagline** – promotional tagline  
- **Keywords** – extracted keyword names  
- **Production Companies** – company names

By concatenating these, each movie gets one unified text field (`tags`) that captures its content, themes, and context.

### 📌 Purpose
- Creates a single feature for text vectorization.  
- Improves similarity search and recommendation models.  
- Simplifies preprocessing for NLP pipelines.


```python
final_df['tags']=final_df['overview']+final_df['genres']+final_df['tagline']+final_df['keywords']+final_df['production_companies']
```

### 🌱 Stemming the Tags Column

I applied **stemming** to reduce words to their root form using NLTK’s `PorterStemmer`.  
This helps normalize variations of words (e.g., *running*, *runs*, *ran* → *run*).

```python
import nltk
```

```python
df2.tags.iloc[1]
```

```python
from nltk import  PorterStemmer
ps =PorterStemmer()
```

We define a helper function `stem(text)` that applies NLTK’s `PorterStemmer` to each word in the input 

```python
def stem(text):
    y=[]
    for i in text.split():
        y.append(ps.stem(i))
    return ' '.join(y)
```

```python
stem('captain barbossa, long believed to be dead, has come back to life and is headed to the edge of the earth with will turner and elizabeth swann. but nothing is quite as it seems.adventure, fantasy, actionat the end of the world, the adventure begins.ocean, drug abuse, exotic island, east india trading company, love of ones life, traitor, shipwreck, strong woman, ship, alliance, calypso, afterlife, fighter, pirate, swashbuckler, aftercreditsstingerwalt disney pictures, jerry bruckheimer films, second mate productions')
```

```python
df2['tags']=df2['tags'].apply(stem)
```

```python
cv= CountVectorizer(max_features=5000,stop_words='english')
```

```python
vectors=cv.fit_transform(df2['tags']).toarray()
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity= cosine_similarity(vectors)
```

```python
def recommend(movie):
    movie_index = df2[df2['original_title'] == movie].index[0]
    distances = similarity[movie_index]
    movie_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:6]
    for i in movie_list:
        print(df2.iloc[i[0]].original_title)
```

```python
recommend("The Dark Knight")
```
