### 📝 Creating the Tags Column

To prepare text data for NLP tasks, we combine multiple descriptive fields into a single `tags` column:

- **Overview** – movie plot summary 

- **Genres** – cleaned genre names  

- **Tagline** – promotional tagline  

- **Keywords** – extracted keyword names  

- **Production Companies** – company names

By concatenating these, each movie gets one unified text field (`tags`) that captures its content, themes, and context.

### 📌 Purpose
- Creates a single feature for text vectorization.  

- Improves similarity search and recommendation models. 
 
- Simplifies preprocessing for NLP pipelines.


```python
final_df['tags']=final_df['overview']+final_df['genres']+final_df['tagline']+final_df['keywords']+final_df['production_companies']
```

# 🧑‍💻 NLP Vectorization Example

This snippet shows how we use **CountVectorizer** and **NearestNeighbors** from scikit-learn to process movie tags and build a similarity model.



```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.neighbors import NearestNeighbors
```

```python
vectorizer = CountVectorizer(stop_words='english')
tag_matrix = vectorizer.fit_transform(final_df['tags'])
```

### 📌 Explanation

CountVectorizer: Transforms text into a bag-of-words representation.

NearestNeighbors: Finds similar movies based on cosine similarity of tag vectors.


```python
knn=NearestNeighbors(metric='cosine',algorithm='brute')
knn.fit(tag_matrix)
```

# 🎬 Movie Recommendation Function

This function uses **k-nearest neighbors (KNN)** to recommend movies based on similarity of tags.

```python
def recommend(movie, n_reco=5, column='original_title'):
    movie_index = final_df[final_df[column] == movie].index[0]
    movie_vector = tag_matrix[movie_index]

    distances, indices = knn.kneighbors(movie_vector, n_neighbors=n_reco+1)

    recommended_indices = indices.flatten()[1:]
    recommended_movies = final_df.iloc[recommended_indices][column]

    print(f"You May Like:")
    for m in recommended_movies:
        print(m)



```

# Example usage

```python
recommend("The Dark Knight Rises")
```
